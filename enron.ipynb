```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
iris = load_iris()
X = iris.data
y = iris.target
```

```python
print("Nama fitur:", iris.feature_names)
print("Nama Target (Spesies):", iris.target_names)
print("Bentuk data X:", X.shape)
print("Bentuk data y:", y.shape)
print("\nBeberapa baris pertama data fitur (X):\n", X[:5])
print("\nBeberapa baris pertama label (y):\n", y[:5])
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

print("\nBentuk X_train:", X_train.shape)
print("Bentuk X_test:", X_test.shape)
print("Bentuk Y_train:", y_train.shape)
print("Bentuk y_test:", y_test.shape)
```

```python
knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X_train, y_train)
print("\nModel KNN telah dilatih.")
```

```python
y_pred = knn.predict(X_test)
cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(6,4))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
            xticklabels=iris.target_names,
            yticklabels=iris.target_names)
plt.xlabel("Prediksi")
plt.ylabel("Aktual")
plt.title("Confusion Matrix KNN pada Dataset Iris")
plt.show()
```

```python
plt.Figure(figsize=(8,6))
sns.scatterplot(x=X[:,0], y=X[:,1], hue=iris.target_names[y], palette="Set1", s=80, edgecolor="k")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Sepal Width (cm)")
plt.title("Sebaran Data Iris berdasarkan Sepal Length & Sepal Width")
plt.legend(title="Spesies")
plt.show()
```

```python
# Membaca data
iris = pd.read_csv("enron_test_fold_1.csv")

# Menghilangkan kolom yang tidak penting
iris.drop("0", axis=1, inplace=True)
```

```python
print(iris.columns.tolist())
print(iris.head())
```

```python
# Kolom kategori/label, contohnya B.B13 atau C.C1
label_cols = [kolom for kolom in iris.columns if "." in kolom]

print(label_cols)  # Periksa dulu apakah yang terpilih memang kolom label

X = iris.drop(columns=label_cols)  # Fitur kata
y = iris[label_cols]               # Label kategori
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.1, random_state=123
)
```

```python
from sklearn.tree import DecisionTreeClassifier
tree_model = DecisionTreeClassifier()
tree_model = tree_model.fit(X_train, y_train)
```

```python
from sklearn.metrics import accuracy_score
y_pred = tree_model.predict(X_test)
acc_secore = round(accuracy_score(y_pred, y_test), 3)
print('Accuracy:', acc_secore)
```

```python
# X_test harus berisi fitur Enron yang dipakai untuk melatih tree_model
prediksi = tree_model.predict(X_test.iloc[[0]])

print(prediksi[0])  # Hasil prediksi untuk baris pertama
```

```python
label_aktif = y_train.columns[prediksi[0] == 1]
print(label_aktif.tolist())
```

```python
import pandas as pd
df = pd.read_csv('enron_test_fold_1.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
data = df.drop(columns=['0'])
data = pd.get_dummies(data)
data
```

```python
print(data.columns[-20:].tolist())
```

```python
import re

# Ambil kolom label dengan pola seperti B.B6, D.D3, C.C12
label_cols = [
    col for col in data.columns
    if re.fullmatch(r"[A-Z]\.[A-Z]\d+", str(col))
]

print(label_cols)  # pastikan kolom yang terpilih memang label

X = data.drop(columns=label_cols)  # fitur
y = data[label_cols]               # semua label
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scaler.fit(X)
scaled_data = scaler.transform(X)
scaled_data = pd.DataFrame(scaled_data, columns= X.columns)
scaled_data.head()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(scaled_data, y, test_size=0.2, random_state=1)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.multiclass import OneVsRestClassifier

model = OneVsRestClassifier(LogisticRegression(max_iter=1000))
model.fit(X_train, y_train)
```

```python
model.score(X_test, y_test)
```
